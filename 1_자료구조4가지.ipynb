1. 리스트

```python
# ==========================================
# 1. 리스트 (List) -> [ ]
# 특징: 순서가 있고(INdex), 수정이 가능하며, 중복을 허용함
# ==========================================
print("--- 1. 리스트 예시 ---")
fruits_list = ["apple", "banana", "cherry", "apple"]  # 중복 허용 ("apple")
```

```python
# 데이터 추가 및 수정 가능
fruits_list.append("orange")
fruits_list[1] = "blueberry"  # 인덱스로 접근하여 수정
```

```python
print(f"리스트 결과: {fruits_list}")
print(f"첫 번째 요소: {fruits_list[0]}")  # 인덱싱 가능
```

2. 튜플

```python
# ==========================================
# 2. 튜플 (Tuple) -> ( )
# 특징: 순서가 있지만(Index), 절대 수정할 수 없음, 중복을 허용함
# 용도: 값이 바뀌면 안 되는 데이터(예: 좌표, 설정값)를 보호할 때 사용
# ==========================================
print("\n--- 2. 튜플 예시 ---")
user_location = (37.5665, 126.9780, 37.5665)  # 중복 허용 (위도, 경도 등)
```

```python
print(f"튜플 결과: {user_location}")
print(f"위도(인덱스 0): {user_location[0]}")  # 인덱싱 가능
```

```python

# 수정 불가, 에러가 발생
user_location[0] = 35.1796
```

3. 셋

```python
# ==========================================
# 3. 셋 / 집합 (Set) -> { }
# 특징: 순서가 없고(Unordered -> 인덱싱 불가), 중복을 절대 허용하지 않음
# 용도: 중복 제거, 집합 연산(합집합, 교집합)
# ==========================================
print("\n--- 3. 셋(집합) 예시 ---")
# 중복된 값들이 들어간 리스트를 셋으로 변환하여 중복 제거
raw_ids = ["user1", "user2", "user1", "user3", "user2"]
unique_ids = set(raw_ids)
```

```python
print(f"원본 리스트: {raw_ids}")
print(f"셋 결과 (중복 자동 제거): {unique_ids}")
```

```python
# 데이터 추가 및 삭제는 가능 (순서가 없으므로 index는 없음)
unique_ids.add("user4")
unique_ids.remove("user1")
print(f"요소 추가 후 셋: {unique_ids}")
```

4. 딕셔너리

```python
# ==========================================
# 4. 딕셔너리 (Dictionary) -> { Key: Value }
# 특징: 키(Key)와 값(Value)의 쌍으로 이루어짐, 키는 중복될 수 없음
# 용도: 사전처럼 특정 '이름'으로 '데이터'를 빠르게 찾을 때 사용
# ==========================================
print("\n--- 4. 딕셔너리 예시 ---")
user_profile = {
    "name": "테스트",
    "age": 25,
    "email": "test@example.com"
}
```

```python
# 키(Key)를 이용해 값(Value)에 접근 및 수정
print(f"사용자 이름: {user_profile['name']}")
```

```python
user_profile["age"] = 26  # 값 수정
user_profile["is_admin"] = True  # 새로운 키-값 쌍 추가
print(f"딕셔너리 결과: {user_profile}")
```
